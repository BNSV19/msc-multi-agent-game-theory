# Importing Required Libraries

```python
import nashpy as nash
import numpy as np
```

# Inputing the Agents Gains

```python
R1 = np.array([[4,7,12],[8,9,7],[6,5,7]])
R2 = np.array([[10,4,5],[7,5,5],[11,4,6]])
```

# Creating the game

```python
game = nash.Game(R1,R2)
```

# Calculating the Nash Equilibrium

```python
eqs = game.support_enumeration()
list(eqs)
```

## From the above output, we can conclude that Robot 1 plays stratergy r2 with 100% probability and Robot plays stratergy r1 with 100% probability.

Since both the outputs are whole numbers, (r2,r1) is a strict nash equilibrium
